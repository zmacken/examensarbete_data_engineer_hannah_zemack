```python
import duckdb
from pathlib import Path

DB_PATH = Path("data/swedavia.duckdb")   # relativt till notebookens arbetsmapp
print(DB_PATH.resolve(), DB_PATH.exists())

con = duckdb.connect(str(DB_PATH), read_only=True)
```

```python
con.sql("SELECT * FROM raw.flights LIMIT 5").show()
```

```python
df = con.sql("""
    SELECT query_date, direction, count(*) AS antal
    FROM raw.flights
    GROUP BY query_date, direction
    ORDER BY query_date, direction
""").df()
df
```

```python
con.close()
```
